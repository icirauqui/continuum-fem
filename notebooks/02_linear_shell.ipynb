# Linear shell solve

A small clamped shell example using the deterministic builder.

```python
from continuum_fem.systems.shell_builder import ShellBuilder
from continuum_fem.fea.shell import LinearShellSolver

shell = ShellBuilder.make_cantilever_plate_problem(nx=6, ny=3, Lx=1.0, Ly=0.25, t=0.01, E=70e9, nu=0.3, total_load_z=-10.0)
result = LinearShellSolver.solve(shell)
print(result.u.shape, float(abs(result.u).max()))
```
