# Linear truss solve

A compact deterministic axial-bar example.

```python
from continuum_fem.systems.truss_builder import TrussBuilder
from continuum_fem.fea.truss import LinearTrussSolver

truss = TrussBuilder.make_dlt_problem(nx=5, ny=2, Lx=1.0, Ly=0.25, gap=0.1, load_vec=[0.0, 0.0, -100.0])
result = LinearTrussSolver.solve(truss)
print(result.u)
```
